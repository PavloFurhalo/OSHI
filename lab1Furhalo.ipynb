7. Визначити функцію, яка приймає на вході масив і транспонує його. Скористайтесь методом reshape. І не забудьте почати з docstrings.

```python
def transpose_array(arr):
    """
    Функція для транспонування масиву за допомогою reshape.
    
    Parameters:
    arr (numpy.ndarray): Масив, який потрібно транспонувати.
    
    Returns:
    numpy.ndarray: Транспонований масив.
    """
    return arr.reshape(arr.shape[::-1])

original_matrix = np.array([[1, 2, 3], [4, 5, 6]])
transposed = transpose_array(original_matrix)
print("Original matrix:\n", original_matrix)
print("Transposed matrix:\n", transposed)
```

19. Перетворити колонку launched у цілочислену (Int64). Перевірити, чи все пройшло, як очікувалось.

```python
dataframe['launched'] = dataframe['launched'].astype('Int64')

print(dataframe['launched'].dtype)
```

20. Видалити колонки, де кількість непустих значень становить менше 30% від загальної кількості записів.

```python
not_null_counter = dataframe.notnull().sum()

rows_amount = len(dataframe)

min_not_null_count = rows_amount * 0.3

useless_columns = not_null_counter[not_null_counter < min_not_null_count].index

dataframe.drop(columns=useless_columns, inplace=True)

print(dataframe.info())
```

21. Очистити датафрейм від дублікатів (дублюючихся рядків) та записів, де атака продовжувалась відʼємну або нульову кількість часу (що фізично неможливо). Не забудьте оновити індекс. Скільки рядків було видалено? Як ви це перевірили у коді?

```python

start_row_amount = len(dataframe)

dataframe['time_start'] = pd.to_datetime(dataframe['time_start'], errors='coerce')
dataframe['time_end'] = pd.to_datetime(dataframe['time_end'], errors='coerce')
dataframe['attack_duration'] = (dataframe['time_end'] - dataframe['time_start']).dt.total_seconds()
dataframe_without_duplicates = dataframe.drop_duplicates()
dataframe_without_duplicates = dataframe_without_duplicates[dataframe_without_duplicates['attack_duration'] > 0]
dataframe_without_duplicates.reset_index(drop=True, inplace=True)

new_row_count = len(dataframe_without_duplicates)
rows_deleted = start_row_amount - new_row_count
print(f"rows deleted: {rows_deleted}")
```

Кількість видалених рядків це різниця між початковою кількістю рядків та кількістю рядків після очищення від дублікатів і іншого мусору.
